```python
import numpy as np
import pandas as pd;
```

```python
df = pd.read_csv("titanic.csv")

# Select only what we need
dfPassengerData = df[['Pclass', 'Sex', 'Fare', 'Survived']]

sex_binary = []

for val in df['Sex']:
    if val == 'male':
        sex_binary.append(0)
    else:
        sex_binary.append(1)

df['Sex_Binary'] = sex_binary

df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Age_Scaled'] = (df['Age'] - df['Age'].mean()) / df['Age'].std()

X = df[['Pclass', 'Sex_Binary', 'Age_Scaled']].to_numpy()
y = df["Survived"].values.reshape(-1, 1)

# Option 2: cleaner
y = df[["Survived"]].values
matrixWeights = np.array([[0.01],[0.5],[0.02]])
bias = 0.0
learning_rate = 0.01
epochs = 1000
```

```python
from scipy.special import expit

# Forward pass
Z = np.dot(X, matrixWeights) + bias
actFunc = expit(Z)

# Sum Squared Error
SumS2Error = np.sum((y - actFunc)**2)

dCda = -2 * (y - actFunc)
dadz = actFunc * (1-actFunc)
#dzdw = X --> Linear Derivative of Z

# Take loss with respect to z.
chainRule = dCda * dadz

# Weight Change
dW = np.dot(X.T, chainRule)
# Bias Change
dB = np.sum(chainRule)

# Update
matrixWeights -= learning_rate * dW
bias -= learning_rate * dB
```
