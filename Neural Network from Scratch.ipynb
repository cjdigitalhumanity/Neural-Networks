```python
import numpy as np
import pandas as pd;
```

```python
#Initialization and Variables

df = pd.read_csv("titanic.csv")

# Select only what we need
dfPassengerData = df[['Pclass', 'Sex', 'Fare', 'Survived']]

sex_binary = []

for val in df['Sex']:
    if val == 'male':
        sex_binary.append(0)
    else:
        sex_binary.append(1)

df['Sex_Binary'] = sex_binary

df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Age_Scaled'] = (df['Age'] - df['Age'].mean()) / df['Age'].std()

X = df[['Pclass', 'Sex_Binary', 'Age_Scaled']].to_numpy()
y = df["Survived"].values.reshape(-1, 1)

#Layer Input Size

inputLayerSize = 3
hiddenLayerSize = 6 
outputLayerSize = 1

#Initialization
y = df[["Survived"]].values

matrixWeightsL1 = np.array([
    [np.random.random() for _ in range(hiddenLayerSize)]
    for _ in range(inputLayerSize)
])

matrixWeightsL2 = np.array([
    [np.random.random() for _ in range(outputLayerSize)]
    for _ in range(hiddenLayerSize)
])

biasL1 = np.zeros((1, hiddenLayerSize))
biasL2 = np.zeros((1, outputLayerSize))

learning_rate = 0.01
epochs = 1000
```

```python
# Activation Functions

#sigmoid
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

#RelU
def relu(x):
    return np.max(0,x)

#SoftPlus
def softPlus(x):
    return np.log(1 + np.exp(x))
```

```python
#NN Passes & Model


#"Forward"agation#
ZL1 = np.dot(X, matrixWeightsL1) + biasL1
actFuncL1 = sigmoid(ZL1)

#Layer 1
ZA = np.dot(actFuncL1, matrixWeightsL2) + biasL2
actFuncL2 = sigmoid(ZA)

# Sum Squared Error
SumS2Error = np.sum((y - actFuncL2)**2)

#Backpropagation#
dCda = -2 * (y - actFuncL2)
dadz = actFuncL2 * (1-actFuncL2)

# Take loss with respect to z.
chainRule = dCda * dadz


#Hidden Layer Backpropagation
hiddenError = np.dot(chainRule, matrixWeightsL2.T)
hiddenGradient = hiddenError * (actFuncL1 * (1 - actFuncL1))

# Weight Changes
dW1 = np.dot(X.T, hiddenGradient)
dW2 = np.dot(actFuncL1.T, chainRule)

# Bias Changes
dB1 = np.sum(hiddenGradient, axis=0, keepdims=True)
dB2 = np.sum(chainRule, axis=0, keepdims=True)

# Update
matrixWeightsL1 -= learning_rate * dW1
biasL1 -= learning_rate * dB1

matrixWeightsL2 -= learning_rate * dW2
biasL2 -= learning_rate * dB2
```
